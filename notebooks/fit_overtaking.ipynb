# Overtaking model — T7, P(pass | pace delta, override)

The 2026 Manual Override is worth far less than DRS (~0.1-0.2 s vs 0.6-0.8 s), so passing
is harder than in 2024. A logistic on the pace delta plus a 2026-era intercept shift is
refit on rounds 1-6; the shift carries that harder-to-pass reality. Opportunities are
followers within the 1.0 s detection gap on a green lap (F-05); lap 1, pit cycles, safety-
car laps, restarts and passes on retiring cars are excluded, each counted.

**Sprint sessions enter the pass extraction.** Unlike the per-race hazard counts (which
exclude sprints because their shorter distance biases per-race event rates), T7 is a
per-opportunity model: a shorter session simply contributes fewer opportunities without
biasing the fit, and 2026 sprint laps are scarce real passing evidence we should not throw
away. Logic lives in `senna.physics.synth.overtaking`.
