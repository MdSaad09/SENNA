# Corner-speed exponent — T2 aero scaling, fitted not assumed

2026 runs about 30% less downforce (AR-08). Apex speed scales with downforce to some
power; the spec's prior is the fourth root (0.25), but the exponent is **fitted** from
matched 2024-vs-2026 apexes on the circuits with both eras, then applied everywhere.
When the matched-corner sample is too thin or the fit lands outside a plausibility gate,
the prior stands and the fallback is flagged, never silent (interpretation flag #1 in
the plan). Logic lives in `senna.physics.synth.aero`.

```python
import argparse
import asyncio
import os

from senna import config

synth_cfg = config.load("synth")
physics_2026 = config.load("physics_2026")
CALIB = config.get(config.load("data_split"), "calibration_rounds")

# the battery needs Postgres + the FastF1 cache; export DATABASE_URL first
assert os.environ.get("DATABASE_URL"), "export DATABASE_URL (see .env.example) before running"
```

```python
from senna.physics.synth import run

args = argparse.Namespace(only="aero", circuit=None, transforms_out=None, physics_2026_out=None)
transforms = asyncio.run(run.run_battery(args))
transforms["globals"]["corner_speed_exponent"]
```
